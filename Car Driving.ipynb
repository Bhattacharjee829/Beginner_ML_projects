```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
df = pd.read_csv('car driving risk analysis.csv')
```

```python
df
```

```python
df.head()
```

```python
x = df[['speed']]
y = df['risk']
```

```python
x
```

```python
plt.scatter(df['speed'], df['risk'], marker = '+', color = 'red')
plt.xlabel('SPEED')
plt.ylabel('RISK')
plt.title('Speed VS Risk')
```

```python
from sklearn.model_selection import train_test_split
```

```python
xtrain, xtest, ytrain, ytest = train_test_split(x,y, test_size =.40, random_state = 5)
```

```python
xtrain
```

```python
ytrain
```

```python
from sklearn.linear_model import LinearRegression
```

```python
reg = LinearRegression()
```

```python
reg.fit(xtrain, ytrain)
```

```python
reg.predict(xtest)
```

```python
ytest
```

```python
plt.plot(df.speed, reg.predict(df[['speed']]))
plt.scatter(df['speed'], df['risk'], marker = '+', color = 'red')
plt.xlabel('SPEED')
plt.ylabel('RISK')
plt.title('Speed VS Risk')
```
